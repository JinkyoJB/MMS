### Init

```python
# 라이브러리 import
from Turntable import Turntable
import numpy as np
import time
from IPython.display import clear_output
RAD2DEG = 180.0 / np.pi
DEG2RAD = np.pi / 180.0
# 1. 초기화 (ID 0번, 192.168.0.X 대역 고정IP 사용해야 하므로 참고)
motor = Turntable(bd_id=0, ip="192.168.0.10", pulses_per_rev=50000) # 50000pulse 당 1바퀴

# 2. 연결
motor.connect(comm_type=0) # 0:TCP, 1:UDP


# 3. 준비 (정보 확인 및 에러 리셋)
motor.check_drive_info()
motor.check_drive_err()
motor.set_servo_on(True) # 서보 온
# motor.set_acceleration(360*DEG2RAD,360*DEG2RAD) 세팅하지 않을 경우 기본 가감속은 3.14 rad/s^2

```

```python
print("상대 각도 이동 테스트")
rpose=-450*DEG2RAD
rvel=180*DEG2RAD
motor.set_acceleration(180*DEG2RAD,180*DEG2RAD) # 각가속/감속 설정 0.5pi rad/s^2
motor.move_inc(rpose, rvel)
```

```python
print("절대 각도 이동 테스트")
rpose=0*DEG2RAD # 목표 Pose : 180도. 라디안 단위 입력 필요
rvel=90*DEG2RAD # 각속도 : 90도/sec
motor.move_abs(rpose, rvel)
while(True):
    time.sleep(0.05)
    clear_output(wait=True)
    pose= motor.getActualPos()
    vel= motor.getActualVel()
    print("Pose : ",pose)
    print("Vel : ",vel)
    if pose==rpose:
        print(pose)
        break
```

```python
# 서보 오프, 연결 종료
motor.set_servo_on(False)
motor.disconnect()
```

```python
# Position 읽기
pose =motor.getActualPos()
print(pose*RAD2DEG)
```

```python
# 현재 Position을 0으로 초기화
motor.clearpos()
```

```python
# 모션 수행 연속 테스트
motor.set_acceleration(360*DEG2RAD,360*DEG2RAD)
pose=720*DEG2RAD
vel=180*DEG2RAD
motor.move_inc(pose, vel)
motor.wait_motion_done()

pose=0*DEG2RAD
vel=360*DEG2RAD
motor.move_abs(pose, vel)
motor.wait_motion_done()

motor.set_acceleration(45*DEG2RAD,720*DEG2RAD)
pose=-720*DEG2RAD
vel=180*DEG2RAD
motor.move_abs(pose, vel)
motor.wait_motion_done()

motor.set_acceleration(720*DEG2RAD,45*DEG2RAD)
pose=720*DEG2RAD
vel=720*DEG2RAD
motor.move_abs(pose, vel)
motor.wait_motion_done()
```
